# Experiment: Collaborative Filtering Recommendation System

**Aim:** To implement user-based and item-based collaborative filtering using cosine similarity and recommend top-N items to a user.



**Libraries:** Pandas, NumPy and Scikit-learn

Run the cells one by one.

## 1. Import libraries and create rating data

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Rows are users and columns are movies
ratings = pd.DataFrame({
    "Movie A": [5, 4, 0, 0, 2],
    "Movie B": [4, 5, 0, 2, 0],
    "Movie C": [0, 0, 5, 4, 4],
    "Movie D": [0, 2, 4, 5, 4],
    "Movie E": [3, 0, 0, 0, 5]
}, index=["User 1", "User 2", "User 3", "User 4", "User 5"])

print("Movie Rating Table:")
print(ratings)
```

**Note:** A value of `0` means that the user has not rated that movie.

## 2. User-based Collaborative Filtering

```python
# Find similarity between users
user_similarity = cosine_similarity(ratings)

user_similarity = pd.DataFrame(
    user_similarity,
    index=ratings.index,
    columns=ratings.index
)

print("User Similarity:")
print(user_similarity.round(2))
```

### Recommend movies to User 1 using similar users

**ADDED (not in the original notebook):** the next original cell uses a variable `user` that is never defined anywhere in the notebook, so it stops with `NameError: name 'user' is not defined`. The cell below defines it. Change `"User 1"` to recommend for another user.

```python
user = "User 1"   # ADDED: the original notebook never defines `user`
```

```python
# Find similar users
similar_users = user_similarity[user].drop(user)

# Calculate recommendation scores
other_users_ratings = ratings.drop(index=user)
scores = other_users_ratings.T.dot(similar_users)

# Select movies not rated by User 1
unrated = ratings.loc[user] == 0
scores = scores[unrated]

# Display top 2 recommendations
print("Top 2 User-based Recommendations:")
print(scores.sort_values(ascending=False).head(2))
```

## 3. Item-based Collaborative Filtering

```python
# Find similarity between movies
item_similarity = cosine_similarity(ratings.T)

item_similarity = pd.DataFrame(
    item_similarity,
    index=ratings.columns,
    columns=ratings.columns
)

print("Movie Similarity:")
print(item_similarity.round(2))
```

### Recommend movies to User 1 using similar movies

```python
# Ratings given by User 1
user_ratings = ratings.loc["User 1"]

# Calculate scores for each movie
scores = item_similarity.dot(user_ratings)

# Do not recommend movies already rated
unrated = user_ratings == 0
scores = scores[unrated]

# Display top 2 recommendations
print("Top 2 Item-based Recommendations:")
print(scores.sort_values(ascending=False).head(2))
```

## Observations

1. Collaborative filtering recommends items using user preferences.
2. User-based filtering finds users with similar ratings.
3. Item-based filtering finds movies with similar rating patterns.
4. Cosine similarity is used to calculate similarity.
5. Top-N recommendations are selected from items the user has not rated.

## Result

User-based and item-based collaborative filtering were implemented using cosine similarity. The top two movie recommendations for User 1 were generated using both methods.
